# 1. Considera el proceso Poisson no homogéneo de tasa $λ(t) = αβt^{β-1}$ con $α > 0$ y $β > 1$.

## (a) Escribe un algoritmo que permita simular las trayectorias de este proceso para $t ∈ [0, 1]$ utilizando el adelgazamiento de un proceso Poisson homogéneo para $α = 10$ y $β = 2, 4, 8, 10$. ¿Qué puedes decir sobre la tasa de aceptación conforme $β$ crece?

```python
import numpy as np

def PPNoHomogeneo(t = 1, alpha = 10, beta = 2, seed = None):
  if seed != None:
    np.random.seed(seed)
  M = alpha * beta
  T = []
  S = 0
  while True:
    S += np.random.exponential(1/M)
    if S > t:
      break
    U = np.random.uniform()
    if U <= (alpha * beta * S**(beta - 1)) / M:
      T.append(S)
  return np.array(T)
```

```python
beta = [2, 4, 8, 10]
X = []

for i, b in enumerate(beta):
    X.append(PPNoHomogeneo(t=1, alpha=10, beta=b))
    print(f'Trayectoria con β = {b}: {X[i]}')
```

La tasa de aceptación esta dada por
$$\frac{\int_0^1 \lambda(t) dt}{M} = \frac{\alpha}{\alpha \beta} = \frac{1}{\beta}.$$
La cual disminuyes cuando beta crece.

## (b) Otra forma de simular este proceso es como sigue: considera $S_1, S_2, ... \overset{\text{iid}}{\sim} \text{Exp}(1)$ y sea $W_n = S_1 + ... + S_n$. De esta manera se pueden construir los tiempos de salto del proceso Poisson no homogéneo como

## $$T_n = \Lambda^{-1}(W_n).$$

## Genera un algoritmo a partir de este procedimiento para generar trayectorias de este proceso para $\alpha = 10$ y $\beta = 2, 4, 8, 10$ y utilízalo para verificar empíricamente que, para un tiempo fijo $t$,

## $$X_t \sim \text{Pois}(\Lambda(t)).$$

```python
def PPNH2(t = 1, alpha = 10, beta= 2, seed = None):
  if seed != None:
    np.random.seed(seed)
  T = []
  W = 0
  while True:
    W += np.random.exponential(1)
    S = (W / alpha) ** (1 / beta)
    if S > t:
      break
    T.append(S)
  return np.array(T)
```

```python
beta = [2, 4, 8, 10]
N = 100000

for b in beta:
  for t in [0.2, 0.4, 0.6, 0.8, 1]:
    Xt = np.zeros(N)
    for i in range(N):
      T = PPNH2(t, 10, b)
      Xt[i] = len(T)
    lam = 10 * t**b
    print(f'β = {b}, t = {t} ---> media ={Xt.mean():.3f}, varianza = {Xt.var():.3f}, Λ(t)= {lam:.3f}')
```

Las varianzas y medias se aproximan bastante bastante

# 2. Considera un proceso de Poisson mixto tal que

# $$\Lambda \sim Ga(\alpha, \beta),$$

# donde $\beta$ denota el parámetro de tasa y, condicionalmente a $\Lambda = \lambda$, $\{N_t : t \geq 0\}$ es un proceso de Poisson homogéneo con intensidad $\lambda$. Implementa un algoritmo para simular trayectorias del proceso. Genera $10000$ realizaciones y estudia empíricamente la distribución de $N_t$ para un tiempo fijo $t$. Obtén su distribución marginal teórica y compara su media y varianza con las obtenidas mediante simulación.

```python
def PPMixto(t = 1, alpha= 1, beta = 1, seed=None):
  if seed != None:
    np.random.seed(seed)
  L = np.random.gamma(alpha, 1 / beta)
  T = []
  S = 0
  while True:
    S += np.random.exponential(1 / L)
    if S > t:
      break
    T.append(S)
  return np.array(T)
```

Calculemos la marginal:

Como $\Lambda=\lambda$, $N_t | \Lambda=\lambda \sim \text{Pois}(\lambda t)$, y $\Lambda$ con densidad $f_{\Lambda} (\lambda) = \frac{\beta^{\alpha}}{\Gamma(\alpha)}\lambda^{\alpha-1}e^{-\lambda \beta}$. Entonces

$$P(N_t=k)=\int_0^{\infty} P(N_t=k|\Lambda=\lambda)\,f_{\Lambda}(\lambda)\,d\lambda=\int_0^{\infty} e^{-\lambda t}\frac{(\lambda t)^k}{k!}\,\frac{\beta^\alpha}{\Gamma(\alpha)}\lambda^{\alpha-1}e^{-\beta\lambda}\,d\lambda = \frac{\Gamma(\alpha+k)}{k!\,\Gamma(\alpha)}\left(\frac{\beta}{\beta+t}\right)^{\alpha}\left(\frac{t}{\beta+t}\right)^{k},$$

es decir, $N_t \sim BinNeg\left(\alpha, \frac{\beta}{\beta+t}\right)$.

# 3. Considera un proceso de saltos de Markov con distribución inicial $\pi_0 = (1/3, 1/3, 1/3)$ y generador infinitesimal

# $$G = \begin{pmatrix} -0.5 & 0.25 & 0.25 \\ 0.4 & -1 & 0.6 \\ 0.1 & 0.1 & -0.2 \end{pmatrix}.$$

## (a) Obtén la matriz de saltos del proceso.

A partir de las tasas de transición podemos calcular las probas de transición, así

$$P = \begin{pmatrix} 0 & \frac{0.25}{0.5} & \frac{0.25}{0.5} \\ \frac{0.4}{1} & 0 & \frac{0.6}{1} \\ \frac{0.1}{0.2} & \frac{0.1}{0.2} & 0 \end{pmatrix} = \begin{pmatrix} 0 & \frac{1}{2} & \frac{1}{2} \\ 0.4 & 0 & 0.6 \\ \frac{1}{2} & \frac{1}{2} & 0 \end{pmatrix}.$$

## (b) Programa un algoritmo que te permita simular trayectorias de este proceso estocástico.

```python
def saltos_markov(t, pi = np.array([1/3, 1/3, 1/3]), seed = None):
  if seed != None:
    np.random.seed(seed)
  estado = np.random.choice(3, p=pi)
  G = np.array([[-0.5, 0.25, 0.25], [0.4, -1, 0.6], [0.1, 0.1, -0.2]])
  P = np.array([[0, 0.5, 0.5], [0.4, 0, 0.6], [0.5, 0.5, 0]])
  estados = []
  S = 0
  T = []
  while True:
    S += np.random.exponential(-1/G[estado, estado])
    if S > t:
      break
    T.append(S)
    estados.append(estado)
    estado = np.random.choice(3, p = P[estado, :])
  return np.array(T), np.array(estados)
```

```python
T, estados = saltos_markov(10)
print(T)
print(estados)
```

## (c) Utiliza tu código para generar una trayectoria del proceso durante un intervalo de tiempo suficientemente largo. Obtén empíricamente la proporción de tiempo que el proceso permanece en cada estado y compárala con la distribución estacionaria teórica, obtenida a partir de

$$\pi G = 0$$

$$\sum_{i=1}^{3} \pi_i = 1.$$

```python
def prop_tiempos(t, pi = (1/3, 1/3, 1/3), seed = None):
  T, estados = saltos_markov(t, pi, seed)
  tiempos = np.zeros(3)
  tiempos[estados[0]] += T[0]
  for k in range(1, len(T)):
    tiempos[estados[k]] += T[k] - T[k-1]
  tiempos[estados[-1]] += t - T[-1]
  prop_tiempos = tiempos / t
  return prop_tiempos
```

```python
prop_tiempos(10000000)
```

Obtengamos la distribucion estacionaria:

$$\pi G = 0, ∑\limits_{i=1}^{3}\pi_i = 1$$

nos da el siguiente sistema de ecuaciones

$$-0.5\pi_1 + 0.4 \pi_2 + 0.1\pi_3 = 0 \tag{1}$$
$$0.25\pi_1 - \pi_2 + 0.1\pi_3 = 0 \tag{2}$$
$$0.25\pi_1 + 0.6 \pi_2 - 0.2\pi_3 = 0 \tag{3}$$
$$\pi_1 + \pi_2 + \pi_3 = 1. \tag{4}$$

Sumando $(2)$ y $(4)$, tenemos

$$1.25\pi_1 + 1.1\pi_3 = 1$$

y depejando $\pi_1$ tenemos

$$\pi_1 = .8 - .88 \pi_3. \tag{5}$$

Ahora sumando $\frac{5}{2}(1)$ y $-\frac{5}{3}(3)$ tenemos

$$(-1.25 - \frac{1.25}{3})\pi_1 + (0.25 + \frac{1}{3})\pi_3 = -\frac{5}{3}\pi_1 + \frac{1.75}{3}\pi_3 = 0 ⇒ -5\pi_1 + 1.75\pi_3 = 0 \tag{6}.$$

Sustituyendo $(5)$ en $(6)$ tenemos

$$-4 + 4.4\pi_3 + 1.75\pi_3 = -4 - 6.15\pi_3 = 0 ⇒ 6.15\pi_3 = 4 \Rightarrow \pi_3 = \frac{4}{6.15} = \frac{80}{123} ≈ 0.6504$$

Sustituyendo en $(5)$ tenemos

$$\pi_1 = \frac{4}{5} - \frac{44}{50}\frac{80}{123} = \frac{28}{123} ≈ 0.22764$$

$$\pi_2 = \frac{15}{123} \approx 0.12195$$

Así podemos ver que los valores obtenidos empiricamente se acercan a los de la distribucion estacionaria, esto pues el proceso es irreducible. Entonces el proceso converge a la distribucion estacionria.
