# 1. Considera una variable aleatoria $X$ con distribución gamma de parámetros $(α, 1)$ suponiendo $α ≥ 1$ y realiza lo siguiente:

## (a) Muestra que $\frac{X}{\beta}$ se distribuye gamma de parámetros (α, β). ¿Qué significa esto para efectos de la simulacián de variables gamma?

Sea $Y = \frac{X}{\beta}$, así $X = Y\beta$ y el Jacobiando es $J = |\beta|$. Entonces por Teorema de Cambio de Variable, sambemos que $f_{\frac{X}{\beta}} (\beta y) f_X(\beta y)\cdot|J| = \frac{\beta(\beta y)^{\alpha -1} e^{-\beta y}}{Γ(\alpha)} = \frac{\beta^{\alpha}}{Γ(\alpha)} y^{\alpha -1}e^{-\beta y}.$ Así $\frac{X}{\beta} \sim Gamma(\alpha,\beta)$

## (b) Para el caso $α < 1$ muestra que si $X$ es $Gamma (1 + α, 1)$ independiente de $U ∼ U(0, 1)$ entonces $XU^{\frac{1}{\alpha}}$ se distribuye $Gamma(α, 1)$.

Como $X$ y $U$ son independientes, entonces $W = U^{\frac{1}{\alpha}}$ y $X$ tambien lo son. Sea $W = U^{\frac{1}{\alpha}}$ entonces $f_W(w) = f_U(w^{\alpha}) |J| = 1 \cdot \alpha w^{\alpha -1}$.

Ahora, sabemos que si $Y = X\cdot W$, así como $0 < W < 1$ entonces con $W = \frac{Y}{X}$ tenemos que $0 < X < Y$. Así $$f_Y(y) = \int_y^{∞}f_X(x)f_W(\frac{y}{x})\frac{1}{x} dx = \int_y^{\infty}\frac{1}{Γ(\alpha +1)} x^{\alpha}e^{-x} \alpha (\frac{y}{x})^{\alpha -1} \frac{1}{x} dx = \frac{\alpha y^{\alpha -1}}{\Gamma(\alpha +1)}\int_y^{\infty}e^{-x}dx = \frac{\alpha y^{\alpha -1}}{\alpha \Gamma(\alpha)} e^{-y} = \frac{y^{\alpha -1}}{\Gamma(\alpha)} e^{-y}.$$

Con lo cual obetenemos que que $XU^{\frac{1}{\alpha}} \sim Gamma(\alpha, 1)$

## (c) Implementa un algoritmo de aceptación y rechazo para simular de una distribución $Gamma$ de parámetros $(α, 1)$ (con $α ≥ 1$) utilizando como distribución instrumental una $Gamma$ de parámetros $(\lfloorα\rfloor, \frac{\lfloor α \rfloor}{α})$.

Sea $X \sim Gamma(\alpha, 1)$, $\kappa = \lfloor\alpha\rfloor$, $\beta = \frac{\lfloor \alpha \rfloor}{\alpha}$ y $Y ~ Gamma (\kappa, \beta)$. Primero obtengamos el valor minimo de $M$, sea
$$g(x) = \frac{f_X(x)}{f_Y(x)} = \frac{\frac{1}{\Gamma(\alpha)}x^{\alpha -1}e^{-x}} {\frac{\beta^{\kappa}}{\Gamma(\kappa)}x^{\kappa -1}e^{-\beta x}} = \frac{\Gamma(\kappa) x^{\alpha - \kappa} e^{-x(1 - \beta)}}  {\beta^{\kappa} \Gamma(\alpha)}.$$

Sea $ c = \frac{\Gamma(\kappa)}{\beta^{\kappa}\Gamma(\alpha)}$, entonces

$$ln(g(x)) = ln(c) + (\alpha - \kappa)ln(x) - x(1 - \beta)$$

Derivando tenemos

$$\frac{d}{dx}ln(g(x)) = \frac{\alpha - \kappa}{x} - (1 -\beta)$$

Igualando a $0$ obtenemos

$$x = \frac{\alpha -\kappa}{1-\beta} = \frac{\alpha -\kappa}{1 -\frac{\kappa}{\alpha}} = \frac{\alpha -\kappa}{\frac{\alpha -\kappa}{\alpha}} = \alpha$$

Sabemos que será maximo, pues $g(0) = 0$ y $\lim\limits_{x \to \infty} g(x) = 0$ y $g(x) > 0$ $∀x\in(0,\infty)$.

Evalunado obtenemos $$g(\alpha) = \frac{\Gamma(k)}{\Gamma(\alpha) \beta^k} \alpha^{\alpha-k} e^{-(1-\beta)\alpha}.$$

Sabemos que si tenemos $i \in (1, \lfloor\alpha\rfloor)$ y $X_i \sim Exp(1)$ independientes, entonces $\sum X_i \sim Gamma(\lfloor\alpha\rfloor, 1)$. Y por (a) sabemos que $\frac{\sum X_i}{\frac{\lfloor \alpha \rfloor}{\alpha}} \sim Gamma(\lfloor \alpha \rfloor, \frac{\lfloor \alpha \rfloor}{\alpha})$

```python
import numpy as np
from scipy.special import gamma
```

```python
def gamma_floor(n, alpha = 1, seed = 29):
  np.random.seed(seed)
  kappa = int(np.floor(alpha))
  X = np.random.exponential(1, size = (kappa, n))
  sum_X = np.sum(X, axis = 0)
  return sum_X/(kappa/alpha)
```

```python
gamma_floor(100, alpha = 3.6)
```

```python
def gamma_a_r(n, alpha = 1, seed = 29):
  kappa = int(np.floor(alpha))
  beta = kappa/alpha
  c = (gamma(kappa))*(alpha**(alpha - kappa))*(np.exp(-(1- beta)*alpha))/(gamma(alpha)*beta**kappa)
  cota_sup = lambda x: gamma(kappa)*(x**(alpha - kappa))*(np.exp(-(1-beta)*x))/(c * beta**kappa * gamma(alpha))
  U = np.random.uniform(0, 1, size = n*int(np.ceil(c)))
  Y = gamma_floor(n*int(np.ceil(c)), alpha = alpha, seed = seed)
  X = np.zeros(n)
  m = 0
  k = 0
  total_iter = 0
  while k < n:
    if m >= len(Y):
      U = np.random.uniform(0, 1, size = (n-k)*int(np.ceil(c)))
      Y = gamma_floor((n-k)*int(np.ceil(c)), alpha = alpha, seed = seed + k)
      m = 0
    if cota_sup(Y[m]) > U[m]:
      X[k] = Y[m]
      k += 1
    m += 1
    total_iter += 1
  return X, total_iter
```

```python
gamma_a_r(100, alpha = 3.6)
```

## (d) Muestra que si $X$ se distribuye $Gamma (α, 1)$ y $Y$ se distribuye $Gamma (β, 1)$, entonces $\frac{X}{X+Y}$ se distribuye $Beta$ de parámetros $(α, β)$.

Usemos los cambios de variable $W = X + Y$ y $V = \frac{X}{X+Y}$, entonces así $X = WV$ y $Y = W(1-V)$, con lo cual $|J| = W$, así,

$$f_{W,V}(w,v) = f_X(wv)\cdot f_Y(w(1-v))\cdot w = \frac{1}{Γ(\alpha)}{(wv)^{\alpha -1}e^{-wv}} \frac{1}{\Gamma(\beta)}(w(1-v))^{\beta -1}e^{-w(1-v)}w = \frac{e^{-w} w^{\alpha + \beta -1} v^{\alpha -1} (1-v)^{\beta -1}} {\Gamma(\alpha)\Gamma(\beta)}$$

Ahora saquemos la marginal,

$$f_V(v) = \int_0^{\infty} \frac{e^{-w} w^{\alpha + \beta -1} v^{\alpha -1} (1-v)^{\beta -1}} {\Gamma(\alpha)\Gamma(\beta)} dw = \frac{v^{\alpha -1} (1-v)^{\beta -1}}{\Gamma(\alpha)\Gamma(\beta)} \int_0^{\infty} e^{-w} w^{\alpha + \beta -1} dw = \frac{\Gamma(\alpha + \beta)}{\Gamma(\alpha)\Gamma(\beta)} v^{\alpha -1} (1-v)^{\beta -1}.$$

Con lo cual $\frac{X}{X+Y} \sim Beta(\alpha, \beta)$

## (e) Utiliza los resultados anteriores y tu código para simular una muestra de tamaño $n = 100$, $1000$, $10000$ de una beta de parámetros $(2.2, 4.7)$. Para cada tamaño de muestra analiza la eficacia del algoritmo en términos del número total de muestras propuestas y el número total de muestras rechazadas. ¿Qué tan eficiente es el algoritmo?

```python
def beta(n, alpha = 1, beta = 1, seed_1 = 29, seed_2 = 30):
    X, total_X = gamma_a_r(n, alpha = alpha, seed = seed_1)
    Y, total_Y = gamma_a_r(n, alpha = beta, seed = seed_2)
    beta = X / (X + Y)
    total = total_X + total_Y
    rechazadas = total - 2*n
    return beta, total, rechazadas
```

# 1. Simular $U, V ∼ U(−1, 1).$
# 2. Si $0 < S = U^2 + V^2 < 1$ hacer

# $$X = U\sqrt{-\frac{2log(S)}{S}}$$
# $$Y = V\sqrt{-\frac{2log(S)}{S}}.$$

# De esta forma al aceptar se tiene que $X$ y $Y$ son variables aleatorias independientes con distribución normal estándar.

## (a) Implementa el algoritmo de Marsaglia y genera una muestra de tamaño $n = 10$, $100$, $1000$, $10000$ de una distribución normal estándar. Muestra que la probabilidad teórica de aceptación del algoritmo es $π/4$ y compara este valor con la proporción de propuestas aceptadas obtenida en tus simulaciones.

```python
def marsaglia(n, seed = 29):
  np.random.seed(seed)
  U = np.random.uniform(-1, 1, size = int(np.ceil(n*4/np.pi)))
  V = np.random.uniform(-1, 1, size = int(np.ceil(n*4/np.pi)))
  S = U**2 + V**2
  X = np.zeros(int(np.ceil(n/2)))
  Y = np.zeros(int(np.ceil(n/2)))
  m = 0
  k = 0
  total_iter = 0
  while k < np.ceil(n/2):
    if m >= len(S):
      U = np.random.uniform(-1, 1, size = int(np.ceil((n-k)*4/np.pi)))
      V = np.random.uniform(-1, 1, size = int(np.ceil((n-k)*4/np.pi)))
      m = 0
    if S[m] < 1:
      X[k] = U[m]*np.sqrt(-2*np.log(S[m])/S[m])
      Y[k] = V[m]*np.sqrt(-2*np.log(S[m])/S[m])
      k += 1
    m +=1
    total_iter += 1
  Z = np.concatenate((X, Y))
  rechazadas = total_iter - int(np.ceil(n/2))
  return Z[:n], total_iter, rechazadas
```

```python
X = [10, 100, 1000, 10000]
Y = []
for i in X:
  Y.append(marsaglia(i)[1:3])
Y
```

Tenemos que el soporte de $(U,V)$, será $[-1, 1]\times[-1,1]$, así el area total será $A = 4$. Ahora los puntos donde se cumple la desigualdad $U^2+V^2<1$, son el circulo unitarion donde su area es $A = \pi$. Con lo cual la probabilidad de aceptacion es de $\frac{\pi}{4}$

## (b) Compara tu algoritmo contra el método de Box-Muller discutido en clase y el algoritmo por default de R para los mismos tamaños de muestra $n = 10$, $100$, $1000$, $10000$. ¿Cuál es más eficiente para cada caso?

```python
import time

def box_muller(n, seed = 29):
    np.random.seed(seed)
    U = np.random.uniform(0, 1, int(np.ceil(n / 2)))
    V = np.random.uniform(0, 1, int(np.ceil(n / 2)))
    r = np.sqrt(-2*np.log(U))
    theta = 2*np.pi *V
    X = r * np.cos(theta)
    Y = r * np.sin(theta)
    Z = np.concatenate((X, Y))
    return Z[:n]
```

```python
X = [10, 100, 1000, 10000]
for n in X:
  t0 = time.perf_counter()
  box_muller(n)
  t_bm = (time.perf_counter() - t0)* 1e6
  t0 = time.perf_counter()
  marsaglia(n)
  t_mp = (time.perf_counter() - t0) * 1e6
  t0 = time.perf_counter()
  np.random.normal(size=n)
  t_np = (time.perf_counter() - t0)* 1e6
  print(f"{n} --->  Box = {t_bm}, Marsaglia = {t_mp}, Numpy = {t_np}")
```

Se ve que el mas eficiente sin duda es el algoritmo de Numpy, los otros dos no son tan rapidos como este. Aunque si los comparamos a estos dos, podemos ver que Box Muller es mas mejor cuando n es grande y Marsaglia es mejor cuando n es pequeña.
